```python
import pandas as pd
from IPython.display import display

df = pd.read_parquet("food_sample.parquet")

print(f"Products in this sample: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")

display(df.head())
```

```python
# List all available columns
for column in df.columns:
    print(column)
```

```python
# Find columns potentially relevant to the meal planner
keywords = [
    "product_name", "ingredient", "allergen",
    "categor", "label", "nutri",
    "carbon", "environment", "ecoscore"
]

relevant_columns = [
    column for column in df.columns
    if any(word in column.lower() for word in keywords)
]

print("Potentially useful columns:")
for column in relevant_columns:
    print(column)

display(df[relevant_columns].head())
```

```python
# Percentage of null values in relevant columns
missing_data = (
    df[relevant_columns]
    .isna()
    .mean()
    .mul(100)
    .round(1)
    .sort_values(ascending=False)
    .rename("Missing (%)")
)

display(missing_data.to_frame())
```

```python
# Inspect one product vertically
display(df[relevant_columns].iloc[0].to_frame(name="Value"))
```

```python
# Search across product-name fields
name_columns = [
    column for column in df.columns
    if "product_name" in column.lower()
]

if name_columns:
    matches = (
        df[name_columns]
        .fillna("")
        .astype(str)
        .apply(
            lambda column: column.str.contains(
                "pasta", case=False, regex=False
            )
        )
        .any(axis=1)
    )

    print(f"Matching products: {matches.sum()}")
    display(df.loc[matches, relevant_columns].head(10))
else:
    print("No product-name columns found.")
```
