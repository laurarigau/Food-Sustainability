```python
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

## Create Initial Recipe Dataset.

```python
recipes_data = [
    {
        "recipe_id": "R001",
        "recipe_name": "Tomato Pasta",
        "servings": 2,
        "dietary_tag": "vegetarian"
    },
    {
        "recipe_id": "R002",
        "recipe_name": "Chicken Rice",
        "servings": 2,
        "dietary_tag": "omnivore"
    },
    {
        "recipe_id": "R003",
        "recipe_name": "Lentil Salad",
        "servings": 2,
        "dietary_tag": "vegan"
    },
    {
        "recipe_id": "R004",
        "recipe_name": "Vegetable Stir Fry",
        "servings": 2,
        "dietary_tag": "vegan"
    },
    {
        "recipe_id": "R005",
        "recipe_name": "Beef Bolognese",
        "servings": 2,
        "dietary_tag": "omnivore"
    }
]

recipes_df = pd.DataFrame(recipes_data)

display(recipes_df)
```

## Create Recipe Ingredients Dataset.

```python
ingredients_data = [
    # Tomato Pasta
    ("R001", "pasta", 180),
    ("R001", "tomato", 250),
    ("R001", "olive oil", 20),
    ("R001", "garlic", 10),

    # Chicken Rice
    ("R002", "chicken", 250),
    ("R002", "rice", 160),
    ("R002", "carrot", 100),
    ("R002", "olive oil", 20),

    # Lentil Salad
    ("R003", "lentils", 200),
    ("R003", "tomato", 150),
    ("R003", "onion", 80),
    ("R003", "olive oil", 20),

    # Vegetable Stir Fry
    ("R004", "broccoli", 200),
    ("R004", "carrot", 150),
    ("R004", "bell pepper", 150),
    ("R004", "rice", 160),

    # Beef Bolognese
    ("R005", "beef", 250),
    ("R005", "pasta", 180),
    ("R005", "tomato", 200),
    ("R005", "onion", 80)
]

ingredients_df = pd.DataFrame(
    ingredients_data,
    columns=["recipe_id", "ingredient", "quantity_g"]
)

display(ingredients_df)
```

## Data Validation

```python
print("Total recipes:", len(recipes_df))
print("Total ingredient entries:", len(ingredients_df))

print("\nMissing recipe values:")
print(recipes_df.isnull().sum())

print("\nMissing ingredient values:")
print(ingredients_df.isnull().sum())

print("\nInvalid quantities:",
      (ingredients_df["quantity_g"] <= 0).sum())

print("\nUnknown recipe IDs:",
      (~ingredients_df["recipe_id"].isin(
          recipes_df["recipe_id"]
      )).sum())

print("\nDuplicated recipe IDs:",
      recipes_df["recipe_id"].duplicated().sum())
```

```python
recipes_df.to_csv("recipes.csv", index=False)

ingredients_df.to_csv(
    "recipe_ingredients.csv",
    index=False
)

print("Datasets saved successfully!")
```

## Ingredient Standardization

```python
# Standardize ingredient names

ingredient_mapping = {
    "tomatoes": "tomato",
    "lentils": "lentil",
    "carrots": "carrot",
    "onions": "onion",
    "potatoes": "potato",
    "bell peppers": "bell pepper"
}

ingredients_df["ingredient_clean"] = (
    ingredients_df["ingredient"]
    .str.lower()
    .str.strip()
    .replace(ingredient_mapping)
)

display(
    ingredients_df[
        ["ingredient", "ingredient_clean"]
    ]
)
```

```python
unique_ingredients = sorted(
    ingredients_df["ingredient_clean"].unique()
)

print("Total unique ingredients:", len(unique_ingredients))

print("\nIngredients:")
for ingredient in unique_ingredients:
    print("-", ingredient)
```

```python
ingredients_df.to_csv(
    "recipe_ingredients.csv",
    index=False
)

print("Updated ingredient dataset saved!")
```
